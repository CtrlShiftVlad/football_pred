```python
import requests, os
from dotenv import load_dotenv


load_dotenv()    
api_key = os.environ["APISPORTS_KEY"] 

url = "https://v3.football.api-sports.io/fixtures"
headers = {"x-apisports-key": api_key}
params = {"league": 39, "season": 2023}   # 39 = Premier League

r = requests.get(url, headers=headers, params=params) # just sends the HTTP get with the header and params we set
data = r.json()
print(data["response"][:2])
```

```python
import pandas as pd

df = pd.json_normalize(data["response"])
df.head()
df.to_csv("testset.csv")
```
